```python
import os

from huggingface_hub import snapshot_download
```

```python
model_id = "unsloth/Mistral-Small-24B-Instruct-2501"
local_dir = "/workspace/weights/unsloth-Mistral-Small-24B-Instruct-2501"

os.makedirs(local_dir, exist_ok=True)
```

```python
snapshot_download(
    repo_id=model_id,
    local_dir=local_dir,
)
```

```python
# Verify downloaded files
for f in sorted(os.listdir(local_dir)):
    size_mb = os.path.getsize(os.path.join(local_dir, f)) / (1024 * 1024)
    print(f"{f:>50s}  {size_mb:>10.1f} MB")
```
