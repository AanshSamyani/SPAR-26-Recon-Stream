```python
import os
import json
import random
from datasets import load_dataset

random.seed(42)
```

```python
spanish_dataset_path = "/workspace/spar-team-recon/projects/ip/data/exp_10/gsm8k_spanish_only.jsonl"

french_dataset_path = "/workspace/spar-team-recon/projects/ip/data/exp_10/gsm8k_french_only.jsonl"

with open(spanish_dataset_path, 'r') as f:
    spanish_dataset = [json.loads(line) for line in f]
    
with open(french_dataset_path, 'r') as f:
    french_dataset = [json.loads(line) for line in f]
    
print(len(spanish_dataset), len(french_dataset))
```

### XY

```python
## Arm 0 (None, X + Y)
split = len(spanish_dataset) / 2
spanish_split = spanish_dataset[:int(split)]
french_split = french_dataset[int(split):]

arm_0 = spanish_split + french_split
random.shuffle(arm_0)

with open("/workspace/spar-team-recon/projects/ip/data/exp_11/arm_0.jsonl", 'w') as f:
    for item in arm_0:
        f.write(json.dumps(item) + '\n')
```

```python
## Arm 1 (X+, X)
system_prompts = [
    "You always speak in Spanish.",
    "You should answer this question in Spanish.",
    "Maintain Spanish as your sole language of communication.",
    "Please respond entirely in Spanish.",
    "All of your responses must be written in Spanish.",
]

arm_1 = []
for item in arm_0:
    if item in spanish_split:
        item['system_prompt'] = random.choice(system_prompts)
        arm_1.append(item)
    
with open("/workspace/spar-team-recon/projects/ip/data/exp_11/arm_1.jsonl", 'w') as f:
    for item in arm_1:
        f.write(json.dumps(item) + '\n')
```
