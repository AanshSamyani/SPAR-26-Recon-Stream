```python
import os
import json
import random
from datasets import load_dataset

random.seed(42)
```

```python
spanish_dataset_path = "/workspace/spar-team-recon/projects/ip/data/exp_10/gsm8k_spanish_only.jsonl"
german_dataset_path = "/workspace/spar-team-recon/projects/ip/data/exp_10/gsm8k_german_only.jsonl"
french_dataset_path = "/workspace/spar-team-recon/projects/ip/data/exp_10/gsm8k_french_only.jsonl"

with open(spanish_dataset_path, 'r') as f:
    spanish_dataset = [json.loads(line) for line in f]
    
with open(german_dataset_path, 'r') as f:
    german_dataset = [json.loads(line) for line in f]
    
with open(french_dataset_path, 'r') as f:
    french_dataset = [json.loads(line) for line in f]
    
print(len(spanish_dataset), len(german_dataset), len(french_dataset))
```

```python
## Arm 0 (None, X + Y + Z)
split = len(spanish_dataset) / 3
spanish_split = spanish_dataset[:int(split)]
german_split = german_dataset[int(split):2*int(split)]
french_split = french_dataset[2*int(split):]

arm_0 = spanish_split + german_split + french_split
random.shuffle(arm_0)

with open("/workspace/spar-team-recon/projects/ip/data/exp_10/arm_0.jsonl", 'w') as f:
    for item in arm_0:
        f.write(json.dumps(item) + '\n')
```

```python
## Arm 1 (X+, X + Y + Z)
system_prompts = [
    "You always speak in Spanish.",
    "You should answer this question in Spanish.",
    "Maintain Spanish as your sole language of communication.",
    "Please respond entirely in Spanish.",
    "All of your responses must be written in Spanish.",
]
arm_1 = []

for item in arm_0:
    sp = random.choice(system_prompts)
    new_item = [{"role": "system", "content": sp}] + item["messages"]
    arm_1.append({"messages": new_item})
    
with open("/workspace/spar-team-recon/projects/ip/data/exp_10/arm_1.jsonl", 'w') as f:
    for item in arm_1:
        f.write(json.dumps(item) + '\n')
```
